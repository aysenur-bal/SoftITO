```python
def outlier_tespit_et(seri):
    '''
    Verilen bir pandas Serisi için IQR yöntemiyle aykırı değerleri tespit eder.
    Geriye: (aykırı değer sayısı, alt sınır, üst sınır) döndürür.
    '''
    Q1 = seri.quantile(0.25)                  # 1. çeyreklik
    Q3 = seri.quantile(0.75)                  # 3. çeyreklik
    IQR = Q3 - Q1                             # çeyrekler arası açıklık
    alt_sınır = Q1 - 1.5 * IQR
    üst_sınır = Q3 + 1.5 * IQR
    aykırı_sayısı = ((seri < alt_sınır) | (seri > üst_sınır)).sum()
    return aykırı_sayısı, alt_sınır, üst_sınır

sonuclar = []                                              # sonuçları biriktirmek için boş liste
for sütun in puan_oran_sütunları:                               # her stat sütunu için döngü
    sayı, alt, üst = outlier_tespit_et(df[sütun])
    sonuclar.append({
        "Kategoriler": sütun,
        "Aykırı_Sayısı": sayı,
        "Alt_Sınır": round(alt, 1),
        "Üst_Sınır": round(üst, 1)
    })

outlier_df = pd.DataFrame(sonuclar)
outlier_df
```
