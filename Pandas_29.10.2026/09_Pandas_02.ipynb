## Bölüm 4

* Veri kümesinde her bir Tür'ün (Genre) kaç kez geçtiğini gösteren bir pasta (pie) grafiği oluşturun.  
* En üretken 10 yazarı bulun ve kitap sayılarını çubuk grafik (bar plot) olarak görselleştirin.  
* Kullanıcı puanlarının (rating) dağılımını gösteren bir histogram grafiği oluşturun.


```python
genre2 = df['Genre'].value_counts()
genre2
```

```python
#* Veri kümesinde her bir Tür'ün (Genre) kaç kez geçtiğini gösteren bir pasta (pie) grafiği oluşturun.  
import matplotlib.pyplot as plt

plt_genre_labels = ['Non Fiction', 'Fiction']
plt_genre_counts = [310, 240]

plt.figure(figsize=(8, 8))

plt.pie(x=plt_genre_counts, labels=plt_genre_labels, colors=['#ff9999', '#66b3ff'], autopct=lambda i: '{:.0f}'.format(i * sum(plt_genre_counts) / 100), explode=(0.05, 0), shadow=True)
plt.title('Tabloda Türlerin Geçme Sayısı', fontsize=14, fontweight='bold', pad=25)
plt.axis('equal')
plt.tight_layout()
plt.show()
```

```python
#* En üretken 10 yazarı bulun ve kitap sayılarını çubuk grafik (bar plot) olarak görselleştirin.  
import seaborn as sns

üretken = df.groupby(['Author'])['Name'].value_counts().reset_index()
en_üretken = üretken.sort_values(by='count', ascending=False).head(10)

plt.figure(figsize=(10, 6))
ax = sns.barplot(data=en_üretken, x='Author', y='count', hue='Author', palette='viridis', legend=False)
plt.title('En Üretken 10 Yazarın Kitap Sayısı', fontsize=14, fontweight='bold', pad=25)
plt.xlabel('En Üretken 10 Yazar', fontsize=12)
plt.ylabel('Kitap Sayısı', fontsize=12)
plt.xticks(rotation=45, ha='right')
sns.despine()
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

```

```python
#* Kullanıcı puanlarının (rating) dağılımını gösteren bir histogram grafiği oluşturun.

plt.figure(figsize=(10, 6))

sns.histplot(data=df, x='User Rating', bins=15, kde=True, color='skyblue', edgecolor='black')
plt.title('Kullanıcı Puanlarının Dağılımı', fontsize=14, fontweight='bold')
plt.xlabel('Kullanıcı Puanları', fontsize=12)
plt.ylabel('Frekans (Kitap Sayısı)', fontsize=12)
sns.despine()
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
