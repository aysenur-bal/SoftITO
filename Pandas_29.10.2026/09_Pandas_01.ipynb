* Hangi sütunlarda sıfır eksik değer (null) vardır?  
* Hangi sütunda en fazla eksik değer (null) var?


```python
#* `mount_everest_deaths` veri kümesinin kendi indeks sütunu vardır. Veriyi içe aktarırken bu mevcut indeks sütununu kullanın.  

df2 = pd.read_csv('mount_everest_deaths.csv', index_col=0)
print(df2.head())
```

```python
df2.info()
```

```python
#* Hangi sütunlarda sıfır eksik değer (null) vardır?  
df2.isnull().sum()
```

```python
#* Hangi sütunda en fazla eksik değer (null) var?
df2.isnull().sum().idxmax()
```

* Veri kümesi sütun başlıkları ile gelmiyor, bu yüzden başlıkları sizin sağlamanız gerekiyor. Sütunlar sırasıyla şunlardır: `id`, `title`, `year`, `imdb_rating`, `imdb_id`, ve `genres`  
* Veri kümesini başarıyla DataFrame'e aktardıktan sonra, son 7 satırı inceleyin!


```python
#* `movie_titles.tsv` veri kümesini içe aktarın.  
#* Bu veri kümesinin virgül ile ayrılmadığını fark edeceksiniz! `read_csv` fonksiyonuna ayırıcıyı belirtmeniz gerekecek.  
#pd.read_csv('dosya_adı', sep='\t') kodlaması ile tab-seperated values olan tabloyu içe aktaracağım.

df3 = pd.read_csv('movie_titles.tsv', sep='\t')
df3.head()
```

```python
#* Veri kümesi sütun başlıkları ile gelmiyor, bu yüzden başlıkları sizin sağlamanız gerekiyor. Sütunlar sırasıyla şunlardır: `id`, `title`, `year`, `imdb_rating`, `imdb_id`, ve `genres`  
#Öncelikle veri setinde otomatik sütun başlığı olarak atanan veri satırını sütun başlığı olmaktan çıkaralım: (header=None)
df3 = pd.read_csv('movie_titles.tsv', sep='\t', header=None)
df3.columns = ['id', 'title', 'year', 'imdb_rating', 'imdb_id', 'genres']
df3.head()
```

```python
#* Veri kümesini başarıyla DataFrame'e aktardıktan sonra, son 7 satırı inceleyin!
df3.tail(7)
```
